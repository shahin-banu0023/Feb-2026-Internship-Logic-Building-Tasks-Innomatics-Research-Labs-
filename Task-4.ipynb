```python
'''Problem 1: Mobile Recharge Validation System
 Problem Statement
Create a function to validate a mobile recharge request.
Rules:
Recharge amount must be ≥ ₹50
Valid plans: 199, 299, 399, 599
User should be allowed to retry if invalid
Real-Time Use
Telecom recharge apps
Payment gateways
Hint
Use function for validation
Use while loop for retry
'''
valid_plans=[199,299,399,599]
def mobile_recharge_validation(amount):
    while(amount<50):
        amount=int(input("Once retry"))
    while(amount>=50):
        if amount in valid_plans:
            print("mobile recharge is valid")
            break
        else:
            print("Invalid plan, Once try!")
            amount=int(input("Once retry"))
amount=int(input("Enter amount"))
mobile_recharge_validation(amount)
```

```python
'''Problem 2: Inventory Reorder Alert System
Problem Statement
Given a dictionary of products and stock quantities:
Print Reorder Alert if stock < 15
Print Stock OK otherwise
Real-Time Use
Warehouse systems
Retail management
Hint
Use function
Use for loop on dictionary
'''
products={
    "Keyboards":18,
    "Mouse":29,
    "Laptops":12,
    "Computer":9,
    "CPU":35,
    "Watches":70,
    "Scanner":7,
    "Printers":10
}
def check_inventory():
    for item,stock in products.items():
        if stock<15:
            print(f"Alert!: Reorder '{item}' (Current stock:{stock})")
        else:
            print(f"Stock OK: '{item}' (Current stock:{stock})")
check_inventory()
```

```python
'''Problem 3: Student Result Processing System
Problem Statement
Create a function that:
Accepts a list of student marks
Calculates average marks
Prints Pass if average ≥ 50 else Fail
Real-Time Use
Online exam systems
Evaluation portals
Hint
Use loop to calculate total
Return result from function
'''
def student_result_system(marks):
    sum=0
    for point in marks:
        sum+=point
    avg_marks=round(sum/len(marks),2)
    
    if avg_marks>=50:
        return f"Student is pass with {avg_marks} marks"
    else:
        return "Student is fail"
result=student_result_system([90,28,39,88,47,58])
print(result)
```

```python
'''Problem 4: Cab Fare Estimator with Retry Option
Problem Statement
Create a cab fare calculator:
Base fare = ₹50
₹12 per km
Peak hour → 25% extra
Allow retry using while loop
Real-Time Use
Ride-hailing apps
Transport systems
Hint
Function for fare calculation
While loop for retry
'''
def cab_fare_calculator(distance,is_peak):
    base_fare=50
    per_KM_rate=12
    total_fare=base_fare+(distance*per_KM_rate)
    if is_peak:
        total_fare=total_fare*1.25
    return total_fare
while True:
    distance=int(input("Enter a distance"))
    peak=input("Is it peak hour(y/n)").lower()
    is_peak=True if peak=='yes' or peak=='y'else False
    fare=cab_fare_calculator(distance,is_peak)
    print(f"Total fare: {fare}")
    retry=input("Calculate another fare(y/n):").lower()
    if retry!='yes'and retry!='y':
        print("Thank you for Cab Fare Calculation")
        break
```

```python
'''Problem 5: Employee Attendance Eligibility Checker
Problem Statement
Create a function that:
Accepts attendance list ("P", "A")def
Calculates attendance percentage
Returns Eligible if ≥ 75% else Not Eligible
Real-Time Use
HR payroll systems
Internship tracking
 Hint
Use loop to count present days
Use return statement
'''
attendece_list=["P","A","P","P","A","P","P","P","P"]
def emp_attendence_eligibility(attendence_list):
    count=0
    for attendence in attendence_list:
        if attendence=="P":
            count+=1
    avg=round(count/len(attendece_list)*100)
    if avg>=75:
        return f"Employee is Eligible with {avg} attendence percentage"
    else:
        return f"Employee is not eligible due to {avg} attendence percentage"
eligibility=emp_attendence_eligibility(attendece_list)
print(eligibility)
```

```python
'''Problem 6: Password Strength Checker
Problem Statement
Create a function to check password strength.
Rules:
Minimum length = 8
Must contain at least one digit
Must contain at least one special character (@#$)
Real-Time Use
Login systems
Security validation
Hint
Use loop over string
Use conditions inside function
'''
special_chars=["@","#","$"]
def check_password(password):
    if len(password)<8:
        return "Weak, the password length must be atleast 8 characters"
    has_digit=False
    has_special_chars=False
    special_chars="@#$"
    for char in password:
        if char.isdigit():
            has_digit=True
        if char in special_chars:
            has_special_chars=True
    if has_digit and has_special_chars:
        return "Strong password"
    elif not has_digit:
        return "Weak, the password must contain atleast one digit"
    else:
        return "Weak, the password must contain atleast one special character(@#$)"
password=input("Enter password")
final_pass=check_password(password)
print(final_pass)
```
