```python
'''Problem 1: Smart Light Controller
Problem Statement
Create a class to control a smart light that can be turned ON or OFF and display its status.
🌍 Real-Time Use
Smart home automation systems
📥 Sample Input
Light Name: Bedroom Light
Action: ON
📤 Expected Output
Bedroom Light is ON
'''
class SmartLight:
    def __init__(self,light_name,action):
        self.light_name=light_name
        self.action=action
    def light_status(self):
        print(f"{self.light_name} is {self.action}")
light=SmartLight("Bedroom Light","ON")
light.light_status()
```

```python
'''Problem 2: Employee ID Card System
 Problem Statement
Create a class that stores employee name, ID, and department and displays ID card details.
Real-Time Use
Corporate HR systems
📥 Sample Input
Employee Name: Rahul
Employee ID: EMP102
Department: IT

📤 Expected Output
Employee ID Card
Name: Rahul
ID: EMP102
Department: IT
'''
class Emp_id_card_system:
    def __init__(self,emp_name,emp_id,emp_dept):
        self.emp_name=emp_name
        self.emp_id=emp_id
        self.emp_dept=emp_dept
    def emp_ID_details(self):
        print("Employee ID Card")
        print(f"Name: {self.emp_name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.emp_dept}")
emp=Emp_id_card_system("Rahul","EMP102","IT")
emp.emp_ID_details()
#emp_obj=Emp_id_card_system("Arjun","EMP105","AI")
```

```python
'''Problem 3: Mobile Contact Record
Problem Statement
Create a class to store contact name and phone number and display contact information.
Real-Time Use
Mobile contact applications
📥 Sample Input
Contact Name: Anita
Phone Number: 9876543210
📤 Expected Output
Contact Saved
Name: Anita
Phone: 9876543210
'''
class Mbl_contact_record:
    def __init__(self,contact_name,phone_num):
        self.contact_name=contact_name
        self.phone_num=phone_num
    def contact_details(self):
        print("Contact Saved")
        print(f"Name: {self.contact_name}")
        print(f"Phone: {self.phone_num}")
contact=Mbl_contact_record("Anita","9876543210")
contact.contact_details()
```

```python
'''Problem 4: Product Price Tag Generator
Problem Statement
Create a class that stores product name and price and prints a formatted price tag.
 Real-Time Use
Retail billing systems
📥 Sample Input
Product Name: Headphones
Price: 2499
📤 Expected Output
Product: Headphones
Price: ₹2499
'''
class Product_price_tag_gen:
    def __init__(self,product_name,price):
        self.product_name=product_name
        self.price=price
    def tag_gen(self):
        print("Product: ",self.product_name)
        print(f"Price: ₹{self.price}")
product=Product_price_tag_gen("Headphones","2499")
product.tag_gen()
```

```python
'''Problem 5: Movie Rating Display System
Problem Statement
Create a class that stores movie name and rating and displays movie details.
Real-Time Use
Streaming platforms
📥 Sample Input
Movie Name: Inception
Rating: 4.8
📤 Expected Output
Movie: Inception
Rating: 4.8 / 5
'''
class Movie_rating_system:
    def __init__(self,movie_name,rating):
        self.movie_name=movie_name
        self.rating=rating
    def movie_details(self):
        print("Movie: ",self.movie_name)
        print("Rating: ",self.rating,"/ 5")
movie_obj=Movie_rating_system("Inception","4.8")
movie_obj.movie_details()
```

```python
'''Problem 6: Delivery Address Manager
Problem Statement
Create a class that stores customer name and delivery address and prints delivery details.
Real-Time Use
Courier and logistics platforms
📥 Sample Input
Customer Name: Suman
Address: Hyderabad
📤 Expected Output
Delivery Details
Customer: Suman
Address: Hyderabad
'''
class Delivery_addrs_mngr:
    def __init__(self,cust_name,delivery_addrs):
        self.cust_name=cust_name
        self.delivery_addrs=delivery_addrs
    def delivery_details(self):
        print("Delivery Details")
        print("Customer: ",self.cust_name)
        print("Address: ",self.delivery_addrs)
delivery=Delivery_addrs_mngr("Suman","Hyderabad")
delivery.delivery_details()
```
